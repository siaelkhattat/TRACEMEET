```
import json, pathlib
root = pathlib.Path(os.environ["TRACEMEET_DATA_ROOT"]) / "data/raw/summre"
for m in sorted(root.iterdir()):
    meta = json.loads((m / "meta.json").read_text())
    print(m.name, meta["split"], len(meta["speakers"]), "speakers", round(meta["duration_s"] / 60, 1), "min")
```
